```
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)
a = np.random.randint(1, 101, 15)
print(a)
print(a.shape, a.dtype, a.size)
big = a[a > 50]
print(big, 'штук', big.size, 'доля', round(big.size / a.size, 2))
sec = np.random.randint(20, 181, 10)
mins = sec / 60
print('сек', sec)
print('среднее', round(float(mins.mean()), 2), 'min', round(float(mins.min()), 2), 'max', round(float(mins.max()), 2))
marks = np.random.randint(1, 6, (4, 3))
print(marks)
print('по студентам', np.round(marks.mean(axis=1), 2))
print('по предметам', np.round(marks.mean(axis=0), 2))

plt.bar(np.arange(1, 16), a)
plt.title('массив из 15 чисел')
plt.tight_layout()
plt.savefig('dz11_bar.png', dpi=120)
```
