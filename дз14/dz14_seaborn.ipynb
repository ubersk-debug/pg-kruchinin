```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('priznaki.csv')
bar = df.groupby('zhanr', as_index=False)['prosrochka'].mean()
sns.barplot(data=bar, x='zhanr', y='prosrochka')
plt.xticks(rotation=30, ha='right')
plt.title('доля просрочки по жанру')
plt.tight_layout()
plt.savefig('dz14_bar.png', dpi=120)
plt.figure()

ch = df.drop_duplicates('chitatel_id')
sns.histplot(ch['vozrast'], bins=8)
plt.title('возраст читателей')
plt.tight_layout()
plt.savefig('dz14_hist.png', dpi=120)
plt.figure()

b = df.groupby('kniga_id', as_index=False).agg(skok=('skok_brali', 'first'), oc=('sred_ocenka_knigi', 'first'))
sns.scatterplot(data=b, x='skok', y='oc')
plt.xlabel('сколько раз брали')
plt.ylabel('средняя оценка')
plt.title('выдачи и оценка книги')
plt.tight_layout()
plt.savefig('dz14_scatter.png', dpi=120)
plt.figure()

heat = pd.crosstab(df['tema_sovpala'], df['prosrochka'])
print(heat)
sns.heatmap(heat, annot=True, cmap='YlOrRd')
plt.title('тема и просрочка')
plt.tight_layout()
plt.savefig('dz14_heat.png', dpi=120)
```
