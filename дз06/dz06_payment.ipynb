```
import pandas as pd
import matplotlib.pyplot as plt

p = pd.read_csv('payment.csv')
cnt = p.groupby('customer_id').size().sort_values(ascending=False)
print('по числу платежей')
print(cnt.head())
freq = (cnt > 30).astype(int)
print('frequent_payer = 1 у', int(freq.sum()))
sm = p.groupby('customer_id')['amount'].sum().sort_values(ascending=False)
print('по сумме')
print(sm.head().round(2))
med = sm.median()
print('медиана', round(float(med), 2))
print('prospective_client = 1 у', int((sm > med).sum()))

top = cnt.head(10)
plt.bar(top.index.astype(str), top.values)
plt.xticks(rotation=45)
plt.title('клиенты по числу платежей')
plt.xlabel('customer_id')
plt.tight_layout()
plt.savefig('dz06_bar.png', dpi=120)
```
