```
import numpy as np
import matplotlib.pyplot as plt

A = np.array([5, 28, 90])
B = np.array([3, 35, 110])
print('A + B', A + B)
print('скалярное', int(np.dot(A, B)))
print('норма A', round(float(np.linalg.norm(A)), 2))
X = np.vstack([A, B])
W = np.array([2, 1, -3])
print('X * W', X @ W)
print('X^T')
print(X.T)
print('отношения A/B', np.round(A / B, 2))
print('отношения разные, строки независимы')

x = np.arange(3)
w = 0.35
plt.bar(x - w / 2, A, w, label='A')
plt.bar(x + w / 2, B, w, label='B')
plt.xticks(x, ['1', '2', '3'])
plt.title('компоненты векторов A и B')
plt.legend()
plt.tight_layout()
plt.savefig('dz01_bar.png', dpi=120)
```
