```
import numpy as np
import matplotlib.pyplot as plt

nA, nB = 100, 100
pA, pB = 0.4, 0.5
p = (40 + 50) / 200
z = (pA - pB) / np.sqrt(p * (1 - p) * (1 / nA + 1 / nB))
print('H0 доли равны, H1 доля B больше')
print('p', p)
print('z', round(float(z), 2))
print('на 5% не значимо, H0 не отвергаем')

plt.bar(['A', 'B'], [pA, pB])
plt.ylim(0, 1)
plt.title('доли переходов')
plt.tight_layout()
plt.savefig('dz03_bar.png', dpi=120)
```
