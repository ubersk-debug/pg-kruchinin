```
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('priznaki.csv')
bar = df.groupby('zhanr', as_index=False)['prosrochka'].mean()
plt.bar(bar['zhanr'], bar['prosrochka'])
plt.xticks(rotation=30, ha='right')
plt.title('доля просрочки по жанру')
plt.tight_layout()
plt.savefig('dz13_bar.png', dpi=120)
plt.figure()

ch = df.drop_duplicates('chitatel_id')
plt.hist(ch['vozrast'], bins=8)
plt.title('возраст читателей')
plt.xlabel('лет')
plt.tight_layout()
plt.savefig('dz13_hist.png', dpi=120)
plt.figure()

b = df.groupby('kniga_id', as_index=False).agg(skok=('skok_brali', 'first'), oc=('sred_ocenka_knigi', 'first'))
plt.scatter(b['skok'], b['oc'])
plt.xlabel('сколько раз брали')
plt.ylabel('средняя оценка')
plt.title('выдачи и оценка книги')
plt.tight_layout()
plt.savefig('dz13_scatter.png', dpi=120)
plt.figure()

heat = pd.crosstab(df['tema_sovpala'], df['prosrochka'])
print(heat)
plt.imshow(heat.values, aspect='auto')
plt.xticks(range(len(heat.columns)), heat.columns)
plt.yticks(range(len(heat.index)), heat.index)
plt.xlabel('просрочка')
plt.ylabel('тема совпала')
plt.title('тема и просрочка')
plt.colorbar()
plt.tight_layout()
plt.savefig('dz13_heat.png', dpi=120)
print('возраст от', int(ch.vozrast.min()), 'до', int(ch.vozrast.max()))
```
