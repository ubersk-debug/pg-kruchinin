```
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

n = 100
pA, pB = 0.4, 0.5
p = 0.45
z = (pA - pB) / np.sqrt(p * (1 - p) * (1 / n + 1 / n))
z_crit = stats.norm.ppf(0.95)
print('H0: piA = piB, H1: piB > piA')
print('z', round(float(z), 2), 'квантиль', round(float(z_crit), 3))
print('1.42 < 1.645, новый дизайн на 5% не подтверждён')
za, zb = 1.645, 0.84
delta = 0.05
n_need = 2 * p * (1 - p) * (za + zb) ** 2 / delta ** 2
mde = (za + zb) * np.sqrt(2 * p * (1 - p) / n)
print('на эффект 0.05 нужно около', round(n_need), 'на группу')
print('при n=100 MDE', round(float(mde), 2))

plt.bar(['старая', 'новая'], [pA, pB])
plt.ylim(0, 1)
plt.title('доля переходов в Товары')
plt.tight_layout()
plt.savefig('dz04_bar.png', dpi=120)
```
