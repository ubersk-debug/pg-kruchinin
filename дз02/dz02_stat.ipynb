```
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

x = np.array([5, 6, 1, 15, 5, 3, 7, 8, 5, 2, 4])
s = np.sort(x)
print('ряд', s)
print('среднее', round(x.mean(), 3))
print('медиана', float(np.median(x)))
vals, cnt = np.unique(x, return_counts=True)
print('мода', int(vals[cnt.argmax()]))
print('дисперсия', round(float(x.var(ddof=1)), 3))
print('СКО', round(float(x.std(ddof=1)), 3))
print('P10', float(np.percentile(x, 10)))
print('P90', float(np.percentile(x, 90)))
se = x.std(ddof=1) / np.sqrt(len(x))
t = stats.t.ppf(0.975, len(x) - 1)
d = t * se
print('t', round(float(t), 3), 'интервал', round(float(x.mean() - d), 2), round(float(x.mean() + d), 2))

plt.hist(x, bins=8)
plt.title('новые пользователи за 11 дней')
plt.xlabel('человек')
plt.tight_layout()
plt.savefig('dz02_hist.png', dpi=120)
```
